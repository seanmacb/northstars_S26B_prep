```python
import h5py
import numpy as np
import matplotlib.pyplot as plt

output_dir = "./data/likelihoods_from_nb/"
deltaz_list = [0.04, 0.0004]
mag_cut_r = 21
criteria = 'impose_threthold_on_sky_area'
threshold = 1.0 # deg^2
num_events = 1

H0_list = []
like_list = []
for deltaz in deltaz_list:
    output_file = f"{output_dir}/likelihood_deltaz_{deltaz}_magcut_{mag_cut_r}_{criteria}_{threshold}_{num_events}events.h5"
    with h5py.File(output_file, 'r') as f:
        H0_array = f['H0'][:]
        like_array = f['likelihood'][:]
        H0_list.append(H0_array)
        like_list.append(like_array)

plt.style.use('~/research/my_plot_style.style')
plt.figure()
for i, (H0_array, like_array) in enumerate(zip(H0_list, like_list)):
    plt.plot(H0_array, like_array, label=fr'$\Delta z={deltaz_list[i]}(1+z)$')
plt.axvline(71, color='gray', linestyle='--', label=r'true')
plt.title(rf'magcutr={mag_cut_r}, $\Delta\Omega \le {threshold}$ deg$^2$, {num_events} event(s)')
plt.xlabel(r'$H_0$ [km/s/Mpc]')
plt.ylabel('Likelihood')
plt.grid()
plt.legend()
plt.show()
```
